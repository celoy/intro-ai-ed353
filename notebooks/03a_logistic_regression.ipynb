The function `plot_boundary` below plots the data, in physical units, and the regions where the model
predicts a female ($\theta^T x > 0$, that is $h_\theta(x) > 1/2$) or a male.

```python
def plot_boundary(x, y, theta, degree, ax=None):
    """Data (physical units) and decision regions of the logistic model with parameters theta."""
    if ax is None:
        ax = plt.gca()
    X_raw = x * std_x + mean_x
    g0 = np.linspace(X_raw[:, 0].min(), X_raw[:, 0].max(), 200)
    g1 = np.linspace(X_raw[:, 1].min(), X_raw[:, 1].max(), 200)
    G0, G1 = np.meshgrid(g0, g1)
    grid = (np.column_stack([G0.ravel(), G1.ravel()]) - mean_x) / std_x
    female = (build_poly(grid, degree) @ theta > 0).reshape(G0.shape)
    ax.contourf(G0, G1, female.astype(float), levels=[-0.5, 0.5, 1.5],
                colors=['tab:blue', 'tab:red'], alpha=0.15)
    ax.scatter(*X_raw[y == 0].T, marker='.', color='tab:blue', label='male')
    ax.scatter(*X_raw[y == 1].T, marker='*', color='tab:red', label='female')
    ax.set_xlabel('bill depth (mm)')
    ax.set_ylabel('body mass (g)')
    ax.legend()
```
